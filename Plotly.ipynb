```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
tips = px.data.tips()
tips
```

```python
import plotly.express as px

df = px.data.gapminder().query("country=='Pakistan'")
print(df.head())

fig = px.line(
    df,
    x='year',
    y='lifeExp'
    
)

fig.show()
```

```python
fig = px.line(
    df,
    x="year",
    y="lifeExp",
    markers = True
)

fig.show()
```

```python
fig = px.line(
    df,
    x="year",
    y="lifeExp",
    title='Life Expectancy of Pakistan'
)

fig.show()
```

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip"
)

fig.show()
```

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="sex"
)

fig.show()
```

```python
fig = px.scatter(
    tips,
    x='total_bill',
    y='tip',
    size = "size"
)
fig.show()
```

```python
fig = px.bar(
    tips,
    x="day",
    y="total_bill"
)

fig.show()
```

```python
fig = px.bar(
    tips,
    x="day",
    y="total_bill",
    color="sex"
)

fig.show()
```

```python
fig = px.bar(
    tips,
    x="total_bill",
    y="day",
    orientation="h"
)

fig.show()
```

```python
fig = px.box(
    tips,
    x="day",
    y="total_bill"
)

fig.show()
```

```python
fig = px.violin(
    tips,
    x="day",
    y="total_bill"
)

fig.show()
```

```python
fig = px.pie(
    names = ['A', 'B', "C"],
    values = [40,30,30]
)

fig.show()
```

```python
fig = px.pie(
    tips,
    names="day"
)

fig.show()
```

```python
fig = px.area(
    df,
    x="year",
    y="lifeExp"
)

fig.show()
```

```python
iris = px.data.iris()

fig = px.scatter_3d(
    iris,
    x="sepal_length",
    y="sepal_width",
    z="petal_length",
    color="species"
)

fig.show()
```

```python
fig = px.sunburst(
    tips,
    path=["sex","day"]
)

fig.show()
```
