```python
import numpy as np

pakistani_cities = [
    "Lahore", "Karachi", "Islamabad", "Rawalpindi", "Faisalabad",
    "Multan", "Peshawar", "Quetta", "Sialkot", "Gujranwala",
    "Hyderabad", "Bahawalpur", "Sargodha", "Sukkur", "Abbottabad",
    "Mardan", "Kasur", "Sheikhupura", "Rahim Yar Khan", "Gujrat",
    "Okara", "Sahiwal", "Jhelum", "Mingora", "Dera Ghazi Khan",
    "Nawabshah", "Larkana", "Khuzdar", "Chiniot", "Muzaffargarh"
]

df["City"] = np.random.choice(pakistani_cities, size=10000)

df.head(10)
```

```python
px.bar(df.groupby('City')['Revenue'].sum().reset_index(),x='City', y='Revenue')
```

```python
avg_order_value = (
    df.groupby('Category')['Revenue']
        .mean()
    .sort_values(ascending=False)
)
print(avg_order_value)
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    df["Revenue"],
    bins=30
)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Number of Orders")

plt.show()
```

```python
product_df = (
    df.groupby("Product")
      .agg({
          "Revenue":"sum",
          "Quantity":"sum"
      })
      .reset_index()
)

fig = px.scatter(
    product_df,
    x="Quantity",
    y="Revenue",
    size="Revenue",
    hover_name="Product",
    title="Product Performance Dashboard"
)

fig.show()
```

```python
import plotly.express as px

sunburst_df = (
    df.groupby(["Category", "Product"])["Revenue"]
      .sum()
      .reset_index()
)

fig = px.sunburst(
    sunburst_df,
    path=["Category", "Product"],
    values="Revenue",
    title="Revenue Contribution by Category and Product"
)

fig.show()
```
