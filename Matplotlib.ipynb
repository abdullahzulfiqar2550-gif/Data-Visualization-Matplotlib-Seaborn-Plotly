```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)

plt.annotate(
    "Highest",
    xy=(4,25)
)

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)

plt.annotate(
    "Peak",
    xy=(4, 25),
    xytext=(2,30),
    arrowprops = {"arrowstyle" : "->"}
)

plt.show()
```

```python
x = [1, 2, 3]
y = [10, 20, 15]

plt.plot(x, y)

for i in range(len(x)):
    plt.annotate(
        y[i],
        (x[i], y[i])
    )

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.fill_between(x,y)

plt.show()
```

```python
x = [1,2,3,4]

y1 = [10,20,15,25]
y2 = [5,10,8,15]

plt.fill_between(x, y1)
plt.fill_between(x, y2)

plt.show()
```
